```python
import tensorflow as tf
import numpy as np
from tensorflow.keras import layers
```

# Load Dataset

```python
text = open(r"C:\Users\sneha\Downloads\english.csv", "r", encoding="utf-8").read()
```

```python
chars = sorted(set(text))
char_to_id = {c: i for i, c in enumerate(chars)}
id_to_char = np.array(chars)

encoded = np.array([char_to_id[c] for c in text])
```

# create training sequence 

```python
seq_len = 100
X, y = [], []

for i in range(len(encoded) - seq_len):
    X.append(encoded[i:i + seq_len])
    y.append(encoded[i + 1:i + seq_len + 1])
X, y = np.array(X), np.array(y)
```

# Build character-level RNN

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(chars), 64),
    tf.keras.layers.SimpleRNN(128, return_sequences=True),
    tf.keras.layers.Dense(len(chars))
])

model.compile(
    optimizer="adam",
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
)
```

```python
model.fit(X, y, batch_size=64, epochs=10)

```

# Generate text

```python
def generate(seed, length=300):
    seed = ''.join(c for c in seed if c in char_to_id)
    result = seed

    for _ in range(length):
        seq = result[-seq_len:]
        x = np.array([[char_to_id[c] for c in seq]])
        pred = model(x, training=False)
        next_id = tf.random.categorical(pred[:, -1, :], 1)[0, 0].numpy()
        result += id_to_char[next_id]

    return result

print(generate("The", 300))
```
